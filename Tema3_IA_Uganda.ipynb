```python
#####I.Realizați o analiză descriptivă a datelor (curățare, structură, statistici generale).

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Încărcarea datelor din fișierele CSV locale
df_consumption = pd.read_csv('consumption_user.csv', low_memory=False)
df_subjects = pd.read_csv('subject_user.csv', low_memory=False)
```

```python
# 2. Structura și Curățarea datelor
# Curățăm numele ingredientelor de caractere speciale (*) și spații inutile
df_consumption['INGREDIENT_CLEAN'] = df_consumption['INGREDIENT_ENG'].str.replace('*', '', regex=False).str.strip().str.capitalize()

# Unim consumul cu datele despre subiecți pentru a avea context (Vârstă, Sex, Zonă)
df_merged = pd.merge(df_consumption, df_subjects[['SUBJECT', 'SEX', 'AGE_YEAR', 'AREA_TYPE']], on='SUBJECT', how='left')
```

```python
# 3. Statistici Generale
print("--- Statistici Generale ale Setului de Date ---")
print(f"Număr total de înregistrări de consum: {len(df_merged)}")
print(f"Număr de subiecți unici: {df_merged['SUBJECT'].nunique()}")
print(f"Număr de ingrediente unice: {df_merged['INGREDIENT_CLEAN'].nunique()}")

# Calculăm media de calorii pe subiect
avg_calories = df_merged.groupby('SUBJECT')['ENERGY_kcal'].sum().mean()
print(f"Media de calorii consumate per subiect: {avg_calories:.2f} kcal")
```

```python
# 4. Vizualizarea Datelor (Analiza Descriptivă)
plt.figure(figsize=(15, 6))

# Subplot 1: Top 10 cele mai consumate alimente
plt.subplot(1, 2, 1)
df_merged['INGREDIENT_CLEAN'].value_counts().head(10).plot(kind='bar', color='skyblue')
plt.title('Top 10 Alimente Consumate')
plt.ylabel('Frecvență')
plt.xticks(rotation=45, ha='right')

# Subplot 2: Distribuția vârstei subiecților
plt.subplot(1, 2, 2)
sns.histplot(df_subjects['AGE_YEAR'], bins=15, kde=True, color='green')
plt.title('Distribuția Vârstei Subiecților')
plt.xlabel('Ani')

plt.tight_layout()
plt.show()
```

```python
##### II. Identificați asocieri frecvente în consumul alimentelor în diverse țări ale lumii, folosind algoritmul alternativ ales.

## Pasul 1:Instalarea librăriei și Pregătirea Tranzacțiilor


# Instalăm librăria necesară
!pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
from mlxtend.preprocessing import TransactionEncoder
import time

# 1. Creăm "coșurile" de cumpărături (Tranzacții)
# Grupăm alimentele consumate de același subiect în aceeași zi
basket = df_merged.groupby(['SUBJECT', 'SURVEY_DAY'])['INGREDIENT_CLEAN'].apply(list).reset_index()
dataset = basket['INGREDIENT_CLEAN'].tolist()
```

```python
# 2. Transformăm datele în format binar (One-Hot Encoding)
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df_bin = pd.DataFrame(te_ary, columns=te.columns_)

print(f"Număr total de tranzacții: {len(dataset)}")
print(f"Număr total de itemi unici: {len(te.columns_)}")
```

```python
#Pasul 2:Analiza Comparativă (Apriori vs FP-Growth)
# Setăm un suport minim (ex: 7% din populație mănâncă acea combinație)
min_sup = 0.07

# --- Testare APRIORI ---
start = time.time()
frequent_itemsets_apriori = apriori(df_bin, min_support=min_sup, use_colnames=True)
apriori_time = time.time() - start

# --- Testare FP-GROWTH ---
start = time.time()
frequent_itemsets_fp = fpgrowth(df_bin, min_support=min_sup, use_colnames=True)
fp_time = time.time() - start

print(f"Timp Apriori: {apriori_time:.4f} secunde")
print(f"Timp FP-Growth: {fp_time:.4f} secunde")
print(f"Factor de accelerare: {apriori_time / fp_time:.2f}x mai rapid FP-Growth")
```

```python
# Pasul3:Pasul 3: Extragerea și Justificarea Regulilor
# Generăm regulile folosind rezultatul de la FP-Growth
rules = association_rules(frequent_itemsets_fp, metric="lift", min_threshold=1.2)

# Sortăm după Lift (cele mai puternice asocieri)
rules = rules.sort_values('lift', ascending=False)

# Afișăm primele 10 reguli pentru interpretare
print("\nTop 10 Reguli de Asociere identificate:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))

# Salvăm rezultatele pentru raport
rules.to_csv('reguli_asociere_fao.csv', index=False)
```

```python
# Filtrarea datelor intr-un tabel

def vizualizeaza_reguli_relevante(df_reguli, top_n=10):
    # Păstrăm doar asocierile cu adevărat puternice (Lift > 1.2)
    reguli_filtrare = df_reguli[(df_reguli['lift'] > 1.2) & (df_reguli['confidence'] > 0.5)].copy()

    # Transformăm formatul tehnic (frozenset) în text normal pentru raport
    reguli_filtrare['antecedents'] = reguli_filtrare['antecedents'].apply(lambda x: ', '.join(list(x)))
    reguli_filtrare['consequents'] = reguli_filtrare['consequents'].apply(lambda x: ', '.join(list(x)))

    # Sortăm după cele mai surprinzătoare asocieri (Lift)
    return reguli_filtrare.sort_values('lift', ascending=False).head(top_n)

# 2. Apelăm funcția și afișăm rezultatul final
top_reguli = vizualizeaza_reguli_relevante(rules)
display(top_reguli[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Crearea graficului de dispersie pentru toate regulile
plt.figure(figsize=(10, 6))
scatter = sns.scatterplot(data=rules, x="support", y="confidence", hue="lift", size="lift", palette="viridis")

# 2. Personalizarea graficului
plt.title('Analiza Regulilor: Suport vs Încredere (Culoare = Lift)', fontsize=15)
plt.xlabel('Suport (Cât de des apar împreună)', fontsize=12)
plt.ylabel('Încredere (Probabilitatea asocierii)', fontsize=12)
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', title='Lift')
plt.grid(True, linestyle='--', alpha=0.6)

# 3. Salvarea imaginii
plt.savefig('scatter_rules.png', bbox_inches='tight', dpi=300)
plt.show()
```

```python
# Grafic de bare pentru Top 10 reguli după Lift
top_10_lift = top_reguli.head(10)
labels = top_10_lift['antecedents'] + ' -> ' + top_10_lift['consequents']

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_lift['lift'], y=labels, palette="magma")
plt.title('Top 10 Reguli după valoarea Lift', fontsize=15)
plt.xlabel('Valoare Lift (Puterea asocierii)', fontsize=12)
plt.ylabel('Regula (Dacă mănâncă A -> atunci mănâncă B)', fontsize=12)
plt.savefig('top_lift_rules.png', bbox_inches='tight', dpi=300)
plt.show()
```
